# Evaluasi Multi-Horizon dan Prediksi Real-Time Irigasi

Notebook ini menggantikan prediksi **1 menit ke depan** dengan prediksi:

- **30 menit**: peringatan dini;
- **60 menit**: horizon utama untuk keputusan irigasi;
- **180 menit**: perencanaan jangka menengah;
- **360 menit**: informasi tambahan dengan ketidakpastian lebih tinggi.

## Prinsip metodologis

1. Target dan model dibuat terpisah untuk setiap horizon.
2. Data dibagi secara kronologis, bukan secara acak.
3. Terdapat `gap` antara data latih dan data uji untuk mencegah kebocoran target.
4. Model dibandingkan dengan **persistence baseline**, yaitu asumsi bahwa kelembapan masa depan sama dengan kelembapan saat ini.
5. Sensor hujan digunakan untuk membaca hujan aktual dan historis. Sensor tersebut tidak dianggap mengetahui hujan yang belum terjadi.
6. Prediksi real-time membutuhkan buffer data historis minimal 360 menit agar fitur lag dan rolling dapat dihitung.
7. Model tidak boleh mengendalikan pompa secara otomatis apabila performanya belum mengalahkan persistence baseline.

> Ambang irigasi pada notebook ini masih berupa parameter sistem. Nilainya harus dikalibrasi berdasarkan jenis tanaman, kedalaman sensor, karakteristik gambut, dan hasil pengujian lapangan.


## Catatan revisi metodologi

Notebook ini telah diperbaiki untuk:

1. Mengubah timestamp sensor berdetik tidak seragam menjadi bucket **1 menit** dengan `resample`, sehingga observasi asli tidak hilang.
2. Membatasi interpolasi hanya pada gap pendek (maksimum 10 menit).
3. Membuat target langsung untuk 30, 60, 180, dan 360 menit ke depan.
4. Menggunakan pembagian data secara kronologis dan **purging gap** sebesar horizon untuk mengurangi kebocoran temporal.
5. Membandingkan model terhadap **persistence baseline**, bukan hanya melaporkan metrik model.
6. Melewati horizon yang datanya belum memenuhi ukuran minimum, bukan menghentikan seluruh notebook.
7. Menyimpan jumlah train/test, periode pengujian, metrik baseline, dan status operasional untuk pelaporan penelitian.

> Ambang kelembapan dan aturan irigasi masih harus divalidasi berdasarkan jenis tanaman, media tanah, kedalaman sensor, dan kalibrasi lapangan. Model belum boleh dipakai sebagai pengendali pompa otomatis hanya berdasarkan satu kali holdout.


```python
# ============================================================
# CELL 1: Import Library dan Konfigurasi
# ============================================================

from pathlib import Path
from datetime import timedelta
import json
import pickle
import warnings

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

warnings.filterwarnings('ignore')

try:
    from xgboost import XGBRegressor
    XGBOOST_AVAILABLE = True
except ImportError:
    XGBOOST_AVAILABLE = False
    print(
        'Peringatan: xgboost belum terpasang. '
        'Notebook tetap berjalan menggunakan Random Forest.'
    )

# ------------------------------------------------------------
# Konfigurasi horizon
# ------------------------------------------------------------
HORIZONS = [30, 60, 180, 360]
PRIMARY_HORIZON = 60

# Sensor yang menjadi target prediksi.
# Ubah menjadi kelembaban_tanah_2 atau kelembaban_tanah_3
# apabila setiap node ingin dimodelkan secara terpisah.
TARGET_BASE_COL = 'kelembaban_tanah_1'

TRAIN_RATIO = 0.80
RANDOM_STATE = 42

# Mode sensor hujan:
# 'auto'   : otomatis mendeteksi data biner atau intensitas;
# 'binary' : 0 = tidak hujan, 1 = hujan;
# 'rate'   : curah_hujan berupa laju mm/jam.
RAIN_MODE = 'auto'

# Ambang operasional sementara
THRESHOLD_KRITIS = 60.0
THRESHOLD_WASPADA = 70.0
THRESHOLD_OPTIMAL = 80.0
THRESHOLD_JENUH = 90.0

DATASET_DIR = Path('dataset')
MODEL_DIR = Path('models')

DATASET_DIR.mkdir(parents=True, exist_ok=True)
MODEL_DIR.mkdir(parents=True, exist_ok=True)

plt.rcParams['figure.dpi'] = 120

print('Horizon prediksi:', HORIZONS)
print('Horizon utama   :', PRIMARY_HORIZON, 'menit')
print('Target sensor   :', TARGET_BASE_COL)
```

```python
# ============================================================
# CELL 2: Load Data Historis
# ============================================================

def load_historical_data():
    """Memuat data historis dari file yang tersedia.

    Prioritas:
    1. dataset/data_clean.csv
    2. dataset/data_cleaned.csv
    3. gabungan dataset/data_train.csv dan dataset/data_test.csv
    """

    candidate_files = [
        DATASET_DIR / 'data_clean.csv',
        DATASET_DIR / 'data_cleaned.csv',
        DATASET_DIR / 'data_cleaning.csv',
    ]

    for file_path in candidate_files:
        if file_path.exists():
            print(f'Menggunakan data historis: {file_path}')
            return pd.read_csv(
                file_path,
                parse_dates=['waktu_pembacaan']
            )

    train_file = DATASET_DIR / 'data_train.csv'
    test_file = DATASET_DIR / 'data_test.csv'

    if train_file.exists() and test_file.exists():
        print(
            'Data bersih tunggal tidak ditemukan. '
            'Menggabungkan data_train.csv dan data_test.csv.'
        )

        df_train_source = pd.read_csv(
            train_file,
            parse_dates=['waktu_pembacaan']
        )
        df_test_source = pd.read_csv(
            test_file,
            parse_dates=['waktu_pembacaan']
        )

        return pd.concat(
            [df_train_source, df_test_source],
            ignore_index=True
        )

    raise FileNotFoundError(
        'Data historis tidak ditemukan. Sediakan salah satu file: '
        'dataset/data_clean.csv, dataset/data_cleaned.csv, atau '
        'pasangan dataset/data_train.csv dan dataset/data_test.csv.'
    )


df_raw = load_historical_data()

print('Jumlah baris awal:', f'{len(df_raw):,}')
print('Jumlah kolom     :', len(df_raw.columns))
print('Periode          :', df_raw['waktu_pembacaan'].min(),
      'sampai', df_raw['waktu_pembacaan'].max())
```

```python
# ============================================================
# CELL 3: Normalisasi Interval Waktu dan Cleaning Dasar (REVISI)
# ============================================================

RAW_SENSOR_COLS = [
    'sensor_tanah_adc_1',
    'kelembaban_tanah_1',
    'kelembaban_udara_1',
    'tekanan_udara_1',
    'suhu_udara_1',
    'sensor_tanah_adc_2',
    'kelembaban_tanah_2',
    'kelembaban_udara_2',
    'tekanan_udara_2',
    'suhu_udara_2',
    'sensor_tanah_adc_3',
    'kelembaban_tanah_3',
    'kelembaban_udara_3',
    'tekanan_udara_3',
    'suhu_udara_3',
    'curah_hujan',
]

available_sensor_cols = [
    col for col in RAW_SENSOR_COLS
    if col in df_raw.columns
]

required_cols = [
    'waktu_pembacaan',
    TARGET_BASE_COL,
    'curah_hujan',
]

missing_required = [
    col for col in required_cols
    if col not in df_raw.columns
]

if missing_required:
    raise KeyError(
        f'Kolom wajib tidak ditemukan: {missing_required}'
    )

df = df_raw[
    ['waktu_pembacaan'] + available_sensor_cols
].copy()

df['waktu_pembacaan'] = pd.to_datetime(
    df['waktu_pembacaan'],
    errors='coerce'
)

df = (
    df.dropna(subset=['waktu_pembacaan'])
      .sort_values('waktu_pembacaan')
)

for col in available_sensor_cols:
    df[col] = pd.to_numeric(
        df[col],
        errors='coerce'
    )

# ------------------------------------------------------------
# PENTING:
# Jangan memakai asfreq('1min') langsung pada timestamp sensor
# yang detiknya berubah-ubah. Data dipetakan dahulu ke bucket
# satu menit agar pembacaan asli tidak hilang.
# ------------------------------------------------------------
df = df.set_index('waktu_pembacaan')

non_rain_cols = [
    col for col in available_sensor_cols
    if col != 'curah_hujan'
]

# Median lebih tahan terhadap spike dan duplikasi dalam menit yang sama.
minute_numeric = (
    df[non_rain_cols]
    .resample('1min')
    .median()
)

# Untuk sensor hujan biner, max menjaga kejadian hujan dalam bucket.
# Untuk sensor laju, median mengurangi spike. Mode awal dideteksi dari data.
raw_rain_non_null = df['curah_hujan'].dropna()
raw_rain_is_binary = (
    raw_rain_non_null.empty
    or raw_rain_non_null.max() <= 1.0
)

if raw_rain_is_binary:
    minute_rain = (
        df['curah_hujan']
        .resample('1min')
        .max()
    )
else:
    minute_rain = (
        df['curah_hujan']
        .resample('1min')
        .median()
    )

df = minute_numeric.join(
    minute_rain.rename('curah_hujan')
)

continuous_cols = non_rain_cols

# Interpolasi hanya untuk gap pendek. Tidak mengisi outage panjang.
df[continuous_cols] = (
    df[continuous_cols]
    .interpolate(
        method='time',
        limit=10,
        limit_direction='both',
        limit_area='inside'
    )
)

# Curah hujan tidak diinterpolasi agar tidak menciptakan hujan palsu.
# NaN pada bucket kosong diasumsikan tidak ada deteksi hujan.
df['curah_hujan'] = (
    df['curah_hujan']
    .fillna(0)
    .clip(lower=0)
)

# Batas fisik dasar
soil_cols = [
    col for col in [
        'kelembaban_tanah_1',
        'kelembaban_tanah_2',
        'kelembaban_tanah_3',
    ]
    if col in df.columns
]

air_humidity_cols = [
    col for col in [
        'kelembaban_udara_1',
        'kelembaban_udara_2',
        'kelembaban_udara_3',
    ]
    if col in df.columns
]

df[soil_cols] = df[soil_cols].clip(0, 100)
df[air_humidity_cols] = df[air_humidity_cols].clip(0, 100)

df = df.reset_index()

interval_minutes = (
    df['waktu_pembacaan']
    .diff()
    .dt.total_seconds()
    .div(60)
)

missing_by_col = (
    df.isna().mean()
    .mul(100)
    .sort_values(ascending=False)
)

print('Jumlah baris setelah resampling:', f'{len(df):,}')
print('Interval dominan:', interval_minutes.mode().iloc[0], 'menit')
print('Missing tersisa:', f'{df.isna().sum().sum():,}')
print()
print('Persentase missing tertinggi:')
print(missing_by_col.head(10).round(2).to_string())

if df[TARGET_BASE_COL].notna().sum() < 1000:
    raise ValueError(
        f'Data valid {TARGET_BASE_COL} terlalu sedikit setelah resampling. '
        'Periksa sumber data dan periode pengambilan.'
    )
```

```python
# ============================================================
# CELL 4: Feature Engineering Multi-Horizon
# ============================================================

def determine_rain_mode(rain_series, requested_mode='auto'):
    if requested_mode in {'binary', 'rate'}:
        return requested_mode

    non_null = rain_series.dropna()

    if non_null.empty:
        return 'binary'

    # Sensor deteksi hujan biasanya hanya berisi 0 dan 1.
    if non_null.max() <= 1.0:
        return 'binary'

    return 'rate'


def build_features(data):
    result = data.copy()
    result = result.sort_values('waktu_pembacaan')

    # --------------------------------------------------------
    # Agregasi tiga node sebagai fitur tambahan
    # --------------------------------------------------------
    soil_available = [
        col for col in [
            'kelembaban_tanah_1',
            'kelembaban_tanah_2',
            'kelembaban_tanah_3',
        ]
        if col in result.columns
    ]

    humidity_available = [
        col for col in [
            'kelembaban_udara_1',
            'kelembaban_udara_2',
            'kelembaban_udara_3',
        ]
        if col in result.columns
    ]

    temperature_available = [
        col for col in [
            'suhu_udara_1',
            'suhu_udara_2',
            'suhu_udara_3',
        ]
        if col in result.columns
    ]

    pressure_available = [
        col for col in [
            'tekanan_udara_1',
            'tekanan_udara_2',
            'tekanan_udara_3',
        ]
        if col in result.columns
    ]

    result['kelembaban_tanah_rata2'] = (
        result[soil_available].mean(axis=1)
    )
    result['kelembaban_udara_rata2'] = (
        result[humidity_available].mean(axis=1)
    )
    result['suhu_udara_rata2'] = (
        result[temperature_available].mean(axis=1)
    )
    result['tekanan_udara_rata2'] = (
        result[pressure_available].mean(axis=1)
    )

    # --------------------------------------------------------
    # Lag dan rolling hanya memakai data saat ini/masa lalu
    # --------------------------------------------------------
    lag_minutes = [1, 5, 15, 30, 60, 120, 180, 360]

    for lag in lag_minutes:
        result[f'{TARGET_BASE_COL}_lag_{lag}m'] = (
            result[TARGET_BASE_COL].shift(lag)
        )

    rolling_windows = [5, 15, 30, 60, 180, 360]

    for window in rolling_windows:
        rolling_target = result[TARGET_BASE_COL].rolling(
            window=window,
            min_periods=window
        )

        result[f'{TARGET_BASE_COL}_mean_{window}m'] = (
            rolling_target.mean()
        )
        result[f'{TARGET_BASE_COL}_std_{window}m'] = (
            rolling_target.std()
        )
        result[f'{TARGET_BASE_COL}_min_{window}m'] = (
            rolling_target.min()
        )
        result[f'{TARGET_BASE_COL}_max_{window}m'] = (
            rolling_target.max()
        )

    for period in [5, 15, 30, 60, 180, 360]:
        result[f'perubahan_target_{period}m'] = (
            result[TARGET_BASE_COL]
            - result[TARGET_BASE_COL].shift(period)
        )

        result[f'laju_perubahan_target_{period}m'] = (
            result[f'perubahan_target_{period}m']
            / period
        )

    # --------------------------------------------------------
    # Fitur hujan aktual dan historis
    # --------------------------------------------------------
    detected_rain_mode = determine_rain_mode(
        result['curah_hujan'],
        RAIN_MODE
    )

    result['status_hujan'] = (
        result['curah_hujan'] > 0
    ).astype(int)

    if detected_rain_mode == 'binary':
        # Jumlah menit hujan dalam setiap jendela.
        rain_interval = result['status_hujan'].astype(float)
        rain_unit = 'menit hujan'
    else:
        # Jika kolom berupa laju mm/jam dan interval satu menit,
        # akumulasi per baris = laju / 60.
        rain_interval = result['curah_hujan'] / 60.0
        rain_unit = 'mm'

    for window in [15, 30, 60, 180, 360]:
        result[f'rain_recent_{window}m'] = (
            rain_interval
            .rolling(window, min_periods=1)
            .sum()
        )

    # --------------------------------------------------------
    # Fitur waktu siklis
    # --------------------------------------------------------
    minute_of_day = (
        result['waktu_pembacaan'].dt.hour * 60
        + result['waktu_pembacaan'].dt.minute
    )

    result['waktu_sin'] = np.sin(
        2 * np.pi * minute_of_day / 1440
    )
    result['waktu_cos'] = np.cos(
        2 * np.pi * minute_of_day / 1440
    )

    day_of_year = result['waktu_pembacaan'].dt.dayofyear

    result['musim_sin'] = np.sin(
        2 * np.pi * day_of_year / 365.25
    )
    result['musim_cos'] = np.cos(
        2 * np.pi * day_of_year / 365.25
    )

    # --------------------------------------------------------
    # Target masa depan yang berbeda untuk setiap horizon
    # --------------------------------------------------------
    for horizon in HORIZONS:
        result[f'target_kelembaban_{horizon}m'] = (
            result[TARGET_BASE_COL].shift(-horizon)
        )

    return result, detected_rain_mode, rain_unit


df_features, detected_rain_mode, rain_unit = build_features(df)

target_cols = [
    f'target_kelembaban_{horizon}m'
    for horizon in HORIZONS
]

excluded_cols = [
    'waktu_pembacaan',
    *target_cols,
]

# Hindari satu sensor yang kosong berat menghapus seluruh baris model.
# Kolom kandidat dipertahankan bila minimal 90% nilainya tersedia.
numeric_candidates = [
    col for col in df_features.columns
    if col not in excluded_cols
    and pd.api.types.is_numeric_dtype(df_features[col])
]

feature_coverage = df_features[numeric_candidates].notna().mean()

feature_cols = [
    col for col in numeric_candidates
    if feature_coverage[col] >= 0.90
]

# Target saat ini wajib tersedia sebagai persistence baseline.
if TARGET_BASE_COL not in feature_cols:
    feature_cols.append(TARGET_BASE_COL)

dropped_low_coverage = [
    col for col in numeric_candidates
    if col not in feature_cols
]

print('Mode sensor hujan terdeteksi:', detected_rain_mode)
print('Satuan fitur hujan historis :', rain_unit)
print('Jumlah fitur model          :', len(feature_cols))
print('Buffer historis minimum     : 360 menit')
print('Fitur dibuang (coverage <90%):', len(dropped_low_coverage))
if dropped_low_coverage:
    print(dropped_low_coverage)
```

## Catatan tentang sensor hujan

Sensor hujan lokal tetap dipakai, tetapi maknanya dibatasi secara benar:

- `status_hujan` menunjukkan hujan yang sedang terdeteksi;
- `rain_recent_60m` dan fitur serupa menunjukkan hujan yang telah terjadi;
- sensor lokal tidak dianggap mengetahui apakah hujan akan turun beberapa jam kemudian.

Karena itu, horizon 360 menit diberi tingkat ketidakpastian yang lebih tinggi. Notebook tidak menggunakan informasi cuaca masa depan sebagai fitur.


```python
# ============================================================
# CELL 5: Fungsi Evaluasi
# ============================================================

def calculate_metrics(y_true, y_pred):
    y_true = np.asarray(y_true, dtype=float).ravel()
    y_pred = np.asarray(y_pred, dtype=float).ravel()

    mae = mean_absolute_error(y_true, y_pred)
    rmse = np.sqrt(mean_squared_error(y_true, y_pred))
    r2 = r2_score(y_true, y_pred)

    denominator = np.maximum(
        np.abs(y_true),
        1e-6
    )

    mape = np.mean(
        np.abs((y_true - y_pred) / denominator)
    ) * 100

    return {
        'MAE': float(mae),
        'RMSE': float(rmse),
        'R2': float(r2),
        'MAPE': float(mape),
    }


def model_is_usable(model_metrics, baseline_metrics):
    """Model dianggap memberi nilai tambah jika mengalahkan persistence.

    Syarat minimum:
    1. RMSE lebih kecil daripada persistence baseline; dan
    2. R² positif.
    """

    return (
        model_metrics['RMSE'] < baseline_metrics['RMSE']
        and model_metrics['R2'] > 0
    )
```

```python
# ============================================================
# CELL 6: Training dan Evaluasi Setiap Horizon (REVISI)
# ============================================================

results = {}
registry = {
    'primary_horizon_minutes': PRIMARY_HORIZON,
    'target_base_column': TARGET_BASE_COL,
    'rain_mode': detected_rain_mode,
    'rain_feature_unit': rain_unit,
    'sampling_interval_minutes': 1,
    'train_ratio': TRAIN_RATIO,
    'horizons': {},
}

MIN_TRAIN_ROWS = 1000
MIN_TEST_ROWS = 200

for horizon in HORIZONS:
    print()
    print('=' * 70)
    print(f'HORIZON {horizon} MENIT')
    print('=' * 70)

    target_col = f'target_kelembaban_{horizon}m'
    required_for_model = feature_cols + [
        target_col,
        'waktu_pembacaan',
    ]

    df_horizon = (
        df_features[required_for_model]
        .dropna(subset=required_for_model)
        .reset_index(drop=True)
    )

    split_index = int(len(df_horizon) * TRAIN_RATIO)

    # Purging gap: target baris train terakhir tidak boleh berada
    # pada/menyeberangi periode test. Karena interval = 1 menit,
    # jumlah baris gap sama dengan horizon menit.
    gap_rows = int(horizon)
    train_end = split_index - gap_rows

    if (
        train_end < MIN_TRAIN_ROWS
        or len(df_horizon) - split_index < MIN_TEST_ROWS
    ):
        print(
            'DILEWATI: data valid belum cukup. '
            f'Total={len(df_horizon):,}, '
            f'train efektif={max(train_end, 0):,}, '
            f'test={max(len(df_horizon)-split_index, 0):,}.'
        )
        continue

    train_data = df_horizon.iloc[:train_end].copy()
    test_data = df_horizon.iloc[split_index:].copy()

    X_train = train_data[feature_cols]
    y_train = train_data[target_col]
    X_test = test_data[feature_cols]
    y_test = test_data[target_col]

    y_pred_persistence = X_test[TARGET_BASE_COL].to_numpy()
    persistence_metrics = calculate_metrics(
        y_test, y_pred_persistence
    )

    rf_model = RandomForestRegressor(
        n_estimators=300,
        max_depth=None,
        min_samples_leaf=2,
        max_features='sqrt',
        n_jobs=-1,
        random_state=RANDOM_STATE
    )
    rf_model.fit(X_train, y_train)
    y_pred_rf = rf_model.predict(X_test)
    rf_metrics = calculate_metrics(y_test, y_pred_rf)

    candidate_models = {
        'Random Forest': {
            'model': rf_model,
            'prediction': y_pred_rf,
            'metrics': rf_metrics,
        }
    }

    if XGBOOST_AVAILABLE:
        xgb_model = XGBRegressor(
            n_estimators=350,
            learning_rate=0.05,
            max_depth=7,
            min_child_weight=3,
            subsample=0.85,
            colsample_bytree=0.85,
            objective='reg:squarederror',
            tree_method='hist',
            random_state=RANDOM_STATE,
            n_jobs=-1
        )
        xgb_model.fit(X_train, y_train)
        y_pred_xgb = xgb_model.predict(X_test)
        xgb_metrics = calculate_metrics(y_test, y_pred_xgb)
        candidate_models['XGBoost'] = {
            'model': xgb_model,
            'prediction': y_pred_xgb,
            'metrics': xgb_metrics,
        }

    best_model_name = min(
        candidate_models,
        key=lambda name: candidate_models[name]['metrics']['RMSE']
    )
    best_entry = candidate_models[best_model_name]
    best_metrics = best_entry['metrics']
    usable = model_is_usable(best_metrics, persistence_metrics)

    operational_status = (
        'LAYAK_SEBAGAI_PENDUKUNG_KEPUTUSAN'
        if usable else 'BELUM_LAYAK_OTOMATIS'
    )

    horizon_suffix = f'{horizon}m'
    best_prefix = 'xgb' if best_model_name == 'XGBoost' else 'rf'
    model_file = MODEL_DIR / f'{best_prefix}_model_{horizon_suffix}.pkl'
    feature_file = MODEL_DIR / f'feature_cols_{horizon_suffix}.json'
    metrics_file = MODEL_DIR / f'metrics_{horizon_suffix}.json'
    prediction_file = DATASET_DIR / f'prediksi_multihorizon_{horizon_suffix}.csv'

    with open(model_file, 'wb') as file:
        pickle.dump(best_entry['model'], file)

    with open(feature_file, 'w', encoding='utf-8') as file:
        json.dump(feature_cols, file, indent=2)

    metrics_payload = {
        'horizon_minutes': horizon,
        'n_train': len(X_train),
        'n_test': len(X_test),
        'gap_rows': gap_rows,
        'test_start': str(test_data['waktu_pembacaan'].min()),
        'test_end': str(test_data['waktu_pembacaan'].max()),
        'persistence': persistence_metrics,
        **{
            name: entry['metrics']
            for name, entry in candidate_models.items()
        },
        'best_model': best_model_name,
        'operational_status': operational_status,
    }

    with open(metrics_file, 'w', encoding='utf-8') as file:
        json.dump(metrics_payload, file, indent=2)

    prediction_output = pd.DataFrame({
        'waktu_pembacaan': test_data['waktu_pembacaan'],
        'kelembaban_sekarang': X_test[TARGET_BASE_COL].to_numpy(),
        'aktual_masa_depan': y_test.to_numpy(),
        'prediksi_persistence': y_pred_persistence,
    })

    for name, entry in candidate_models.items():
        output_name = 'prediksi_xgb' if name == 'XGBoost' else 'prediksi_rf'
        prediction_output[output_name] = entry['prediction']

    prediction_output.to_csv(prediction_file, index=False)

    results[horizon] = {
        'target_col': target_col,
        'test_data': test_data,
        'X_test': X_test,
        'y_test': y_test,
        'persistence_metrics': persistence_metrics,
        'candidate_models': candidate_models,
        'best_model_name': best_model_name,
        'best_metrics': best_metrics,
        'operational_status': operational_status,
        'model_file': str(model_file),
        'feature_file': str(feature_file),
        'prediction_file': str(prediction_file),
    }

    registry['horizons'][str(horizon)] = {
        'best_model': best_model_name,
        'model_file': str(model_file),
        'feature_file': str(feature_file),
        'target_column': target_col,
        'operational_status': operational_status,
        'best_metrics': best_metrics,
        'persistence_metrics': persistence_metrics,
    }

    print('Train:', f'{len(X_train):,}')
    print('Test :', f'{len(X_test):,}')
    print('Gap  :', gap_rows, 'baris/menit')
    print('Persistence RMSE:', f'{persistence_metrics["RMSE"]:.4f}')
    print('Model terbaik   :', best_model_name)
    print('Model RMSE      :', f'{best_metrics["RMSE"]:.4f}')
    print('Model R²        :', f'{best_metrics["R2"]:.4f}')
    print('Status          :', operational_status)

if not results:
    raise ValueError(
        'Tidak ada horizon yang dapat dilatih. '
        'Periksa missing values, periode data, dan konfigurasi fitur.'
    )

registry_file = MODEL_DIR / 'model_registry_multihorizon.json'
with open(registry_file, 'w', encoding='utf-8') as file:
    json.dump(registry, file, indent=2)

print()
print('Registry model disimpan:', registry_file)
```

```python
# ============================================================
# CELL 7: Ringkasan Performa Antar-Horizon
# ============================================================

summary_rows = []

for horizon, result in results.items():
    baseline = result['persistence_metrics']

    for model_name, entry in result['candidate_models'].items():
        metrics = entry['metrics']

        summary_rows.append({
            'Horizon_menit': horizon,
            'Model': model_name,
            'MAE': metrics['MAE'],
            'RMSE': metrics['RMSE'],
            'R2': metrics['R2'],
            'MAPE': metrics['MAPE'],
            'RMSE_persistence': baseline['RMSE'],
            'Mengalahkan_persistence': (
                metrics['RMSE'] < baseline['RMSE']
            ),
            'Status_operasional': (
                result['operational_status']
                if model_name == result['best_model_name']
                else 'MODEL_PEMBANDING'
            ),
        })

summary_metrics = pd.DataFrame(summary_rows)

print(
    summary_metrics
    .sort_values(['Horizon_menit', 'RMSE'])
    .round(4)
    .to_string(index=False)
)

summary_metrics.to_csv(
    DATASET_DIR / 'ringkasan_metrik_multihorizon.csv',
    index=False
)
```

```python
# ============================================================
# CELL 10: Logika Rekomendasi Irigasi
# ============================================================

def rekomendasi_irigasi_multihorizon(
    kelembaban_sekarang,
    prediksi,
    horizon_minutes,
    laju_perubahan_60m,
    status_hujan,
    rain_recent_60m,
    rain_mode,
    model_status
):
    """Menghasilkan rekomendasi berbasis kondisi aktual dan forecast.

    Model yang belum mengalahkan persistence baseline hanya boleh
    digunakan sebagai informasi monitoring, bukan pengendali pompa.
    """

    if model_status != 'LAYAK_SEBAGAI_PENDUKUNG_KEPUTUSAN':
        return (
            'MONITORING SAJA',
            'Model belum mengalahkan baseline secara memadai'
        )

    rain_detected = int(status_hujan) == 1

    if rain_mode == 'binary':
        recent_rain_significant = rain_recent_60m >= 5
    else:
        recent_rain_significant = rain_recent_60m >= 1.0

    if rain_detected:
        return (
            'TUNDA IRIGASI',
            'Sensor mendeteksi hujan sedang berlangsung'
        )

    if recent_rain_significant:
        return (
            'EVALUASI ULANG',
            'Hujan baru terjadi; periksa respons kelembapan tanah'
        )

    if kelembaban_sekarang < THRESHOLD_KRITIS:
        return (
            'IRIGASI SEGERA',
            'Kelembapan saat ini sudah berada di bawah ambang kritis'
        )

    if prediksi < THRESHOLD_KRITIS:
        if horizon_minutes <= 60:
            return (
                'SIAPKAN IRIGASI',
                f'Diprediksi kritis dalam {horizon_minutes} menit'
            )

        return (
            'JADWALKAN PEMERIKSAAN',
            f'Diprediksi kritis dalam {horizon_minutes} menit'
        )

    if (
        prediksi < THRESHOLD_WASPADA
        or laju_perubahan_60m < -0.05
    ):
        return (
            'WASPADA',
            'Kelembapan diprediksi rendah atau sedang menurun'
        )

    if prediksi >= THRESHOLD_JENUH:
        return (
            'HENTIKAN IRIGASI',
            'Risiko kondisi terlalu jenuh'
        )

    if prediksi >= THRESHOLD_OPTIMAL:
        return (
            'PERIKSA DRAINASE',
            'Kelembapan relatif tinggi'
        )

    return (
        'NORMAL',
        'Irigasi belum diperlukan'
    )
```

## Prediksi real-time

Prediksi real-time di bawah ini tidak memakai satu baris sensor saja. Sistem mengambil **buffer historis minimal 360 menit** untuk menghitung:

- lag kelembapan;
- rata-rata bergerak;
- laju perubahan;
- durasi atau akumulasi hujan;
- kondisi suhu, kelembapan udara, dan tekanan.

Pada implementasi AIoT, bagian `df_features` dapat diganti dengan hasil query database yang memuat sekurang-kurangnya enam jam data terbaru.


```python
# ============================================================
# CELL 11: Fungsi Prediksi Real-Time Multi-Horizon
# ============================================================

def prediksi_realtime_multi_horizon(
    feature_history,
    registry_path=MODEL_DIR / 'model_registry_multihorizon.json'
):
    if not registry_path.exists():
        raise FileNotFoundError(
            f'Registry model tidak ditemukan: {registry_path}'
        )

    with open(
        registry_path,
        'r',
        encoding='utf-8'
    ) as file:
        model_registry = json.load(file)

    required_buffer = 360
    if len(feature_history) < required_buffer:
        raise ValueError(
            'Prediksi real-time membutuhkan buffer historis '
            f'minimal {required_buffer} baris/menit.'
        )

    output_rows = []

    latest_time = (
        feature_history['waktu_pembacaan'].max()
    )

    for horizon_text, config in model_registry['horizons'].items():
        horizon = int(horizon_text)

        feature_file = Path(
            config['feature_file']
        )
        model_file = Path(
            config['model_file']
        )

        with open(
            feature_file,
            'r',
            encoding='utf-8'
        ) as file:
            model_features = json.load(file)

        valid_history = (
            feature_history
            .dropna(subset=model_features)
        )

        if valid_history.empty:
            raise ValueError(
                f'Tidak ada baris fitur lengkap untuk horizon '
                f'{horizon} menit.'
            )

        latest_row = valid_history.iloc[-1]
        X_latest = pd.DataFrame(
            [latest_row[model_features].to_dict()]
        )

        with open(model_file, 'rb') as file:
            model = pickle.load(file)

        prediction = float(
            np.asarray(
                model.predict(X_latest)
            ).ravel()[0]
        )

        current_soil = float(
            latest_row[TARGET_BASE_COL]
        )

        status_hujan = int(
            latest_row.get('status_hujan', 0)
        )

        rain_recent_60m = float(
            latest_row.get('rain_recent_60m', 0)
        )

        rate_change_60m = float(
            latest_row.get(
                'laju_perubahan_target_60m',
                0
            )
        )

        recommendation, reason = (
            rekomendasi_irigasi_multihorizon(
                kelembaban_sekarang=current_soil,
                prediksi=prediction,
                horizon_minutes=horizon,
                laju_perubahan_60m=rate_change_60m,
                status_hujan=status_hujan,
                rain_recent_60m=rain_recent_60m,
                rain_mode=model_registry['rain_mode'],
                model_status=config['operational_status'],
            )
        )

        uncertainty = (
            'rendah'
            if horizon <= 60
            else 'sedang'
            if horizon <= 180
            else 'tinggi'
        )

        output_rows.append({
            'waktu_sensor': latest_time,
            'berlaku_untuk_waktu': (
                latest_time
                + timedelta(minutes=horizon)
            ),
            'horizon_menit': horizon,
            'kelembaban_sekarang': current_soil,
            'prediksi_kelembaban': prediction,
            'model': config['best_model'],
            'status_model': config['operational_status'],
            'ketidakpastian_relatif': uncertainty,
            'rekomendasi': recommendation,
            'alasan': reason,
        })

    return (
        pd.DataFrame(output_rows)
        .sort_values('horizon_menit')
        .reset_index(drop=True)
    )


hasil_realtime = prediksi_realtime_multi_horizon(
    df_features
)

print('=== PREDIKSI REAL-TIME MULTI-HORIZON ===')
print(
    hasil_realtime[
        [
            'waktu_sensor',
            'berlaku_untuk_waktu',
            'horizon_menit',
            'kelembaban_sekarang',
            'prediksi_kelembaban',
            'model',
            'ketidakpastian_relatif',
            'rekomendasi',
        ]
    ]
    .round(2)
    .to_string(index=False)
)
```

```python
# ============================================================
# CELL 12: Simpan Hasil Real-Time dan Kesimpulan
# ============================================================

realtime_output_file = (
    DATASET_DIR
    / 'hasil_prediksi_realtime_multihorizon.csv'
)

hasil_realtime.to_csv(
    realtime_output_file,
    index=False
)

print('=' * 72)
print('KESIMPULAN EVALUASI MULTI-HORIZON')
print('=' * 72)

for horizon in sorted(results):
    result = results[horizon]

    print()
    print(f'Horizon {horizon} menit')
    print('  Model terbaik :', result['best_model_name'])
    print(
        '  RMSE model    :',
        f'{result["best_metrics"]["RMSE"]:.4f}'
    )
    print(
        '  RMSE baseline :',
        f'{result["persistence_metrics"]["RMSE"]:.4f}'
    )
    print(
        '  R² model      :',
        f'{result["best_metrics"]["R2"]:.4f}'
    )
    print(
        '  Status        :',
        result['operational_status']
    )

print()
print('Horizon utama:', PRIMARY_HORIZON, 'menit')
print('File hasil real-time:', realtime_output_file)
print(
    'Keputusan pompa otomatis hanya boleh diaktifkan '
    'untuk model berstatus LAYAK_SEBAGAI_PENDUKUNG_KEPUTUSAN.'
)
```
