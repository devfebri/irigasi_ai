# 📡 Notebook 1: Preprocessing Data Sensor IoT
**Tujuan:** Membersihkan data mentah dari sensor ESP32 (3 node) dan mempersiapkannya untuk pemodelan Machine Learning.

**Alur:**
1. Load data dari database MySQL
2. Analisis kualitas data (missing values, outlier, nilai 0 tidak valid)
3. Cleaning & Imputasi
4. Sortir berdasarkan waktu
5. Simpan dataset bersih ke CSV

```python
# ============================================================
# CELL 1: Import Library
# ============================================================

import sys
print(sys.executable)

%pip install seaborn sqlalchemy pymysql

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sqlalchemy import create_engine
import warnings
warnings.filterwarnings('ignore')

# Pengaturan tampilan
pd.set_option('display.max_columns', None)
pd.set_option('display.float_format', '{:.4f}'.format)
plt.style.use('seaborn-v0_8-whitegrid')

print('✅ Library berhasil diimport')
print(f'   - pandas  : {pd.__version__}')
print(f'   - numpy   : {np.__version__}')
```

```python
# ============================================================
# CELL 2: Koneksi Database & Load Data
# ============================================================
engine = create_engine(
    "mysql+pymysql://u832389661_unama1:Unama2026@153.92.15.62:3306/u832389661_unama1"
)

query = "SELECT * FROM sensor_readings ORDER BY waktu_pembacaan ASC"
df_raw = pd.read_sql(query, engine)

print('=== 📊 Informasi Data Mentah ===')
print(f'Jumlah baris     : {df_raw.shape[0]:,}')
print(f'Jumlah kolom     : {df_raw.shape[1]}')
print(f'Periode data     : {df_raw["waktu_pembacaan"].min()} s.d. {df_raw["waktu_pembacaan"].max()}')
print(f'Tipe data        :')
print(df_raw.dtypes)
```

```python
# ============================================================
# CELL 3: Analisis Missing Values
# ============================================================
print('=== 🔍 Analisis Missing Values ===')
missing = df_raw.isnull().sum()
missing_pct = (missing / len(df_raw) * 100).round(3)

missing_df = pd.DataFrame({
    'Jumlah_Missing': missing,
    'Persentase(%)': missing_pct
})
print(missing_df[missing_df['Jumlah_Missing'] > 0])
print(f'\nTotal kolom tanpa missing value: {(missing == 0).sum()}')
```

```python
# ============================================================
# CELL 4: Analisis Nilai Nol (Tidak Valid)
# ============================================================
print('=== ⚠️  Analisis Nilai Nol (Berpotensi Tidak Valid) ===')
sensor_cols = [
    'kelembaban_tanah_1', 'kelembaban_tanah_2', 'kelembaban_tanah_3',
    'kelembaban_udara_1', 'kelembaban_udara_2', 'kelembaban_udara_3',
    'suhu_udara_1', 'suhu_udara_2', 'suhu_udara_3',
    'tekanan_udara_1', 'tekanan_udara_2', 'tekanan_udara_3',
    'curah_hujan'
]

zero_analysis = pd.DataFrame({
    'Jumlah_Nol': (df_raw[sensor_cols] == 0).sum(),
    'Persentase(%)': ((df_raw[sensor_cols] == 0).sum() / len(df_raw) * 100).round(3)
})
print(zero_analysis)

print('\n=== ⚠️  Analisis Nilai Negatif ===')
for col in ['tekanan_udara_1', 'tekanan_udara_2', 'tekanan_udara_3']:
    neg_count = (df_raw[col] < 0).sum()
    print(f'{col}: {neg_count} nilai negatif')
```

```python
# ============================================================
# CELL 5: Visualisasi Distribusi Data Mentah
# ============================================================
fig, axes = plt.subplots(3, 3, figsize=(16, 12))
fig.suptitle('📊 Distribusi Kelembaban Tanah & Sensor per Node', fontsize=14, fontweight='bold')

target_cols = ['kelembaban_tanah_1', 'kelembaban_tanah_2', 'kelembaban_tanah_3']
udara_cols  = ['kelembaban_udara_1', 'kelembaban_udara_2', 'kelembaban_udara_3']
suhu_cols   = ['suhu_udara_1', 'suhu_udara_2', 'suhu_udara_3']

colors = ['#2196F3', '#4CAF50', '#FF9800']
for i, col in enumerate(target_cols):
    axes[0, i].hist(df_raw[col], bins=50, color=colors[i], alpha=0.7, edgecolor='white')
    axes[0, i].set_title(f'Kelembaban Tanah Node {i+1}', fontweight='bold')
    axes[0, i].set_xlabel('Kelembaban (%)')

for i, col in enumerate(udara_cols):
    axes[1, i].hist(df_raw[col], bins=50, color=colors[i], alpha=0.7, edgecolor='white')
    axes[1, i].set_title(f'Kelembaban Udara Node {i+1}', fontweight='bold')
    axes[1, i].set_xlabel('Kelembaban (%)')

for i, col in enumerate(suhu_cols):
    axes[2, i].hist(df_raw[col], bins=50, color=colors[i], alpha=0.7, edgecolor='white')
    axes[2, i].set_title(f'Suhu Udara Node {i+1}', fontweight='bold')
    axes[2, i].set_xlabel('Suhu (°C)')

plt.tight_layout()
plt.savefig('dataset/plot_distribusi_raw.png', dpi=150, bbox_inches='tight')
plt.show()
print('✅ Plot disimpan ke dataset/plot_distribusi_raw.png')
```

```python
# ============================================================
# CELL 6: Cleaning Data
# ============================================================
df = df_raw.copy()

print('=== 🧹 Proses Cleaning Data ===')
print(f'Baris awal: {len(df):,}')

# --- 6a. Convert waktu_pembacaan ke datetime dan set sebagai index ---
df['waktu_pembacaan'] = pd.to_datetime(df['waktu_pembacaan'])
df = df.sort_values('waktu_pembacaan').reset_index(drop=True)

# --- 6b. Hapus kolom yang tidak dibutuhkan untuk ML ---
drop_cols = ['id', 'backup_status', 'created_at', 'updated_at']
df = df.drop(columns=drop_cols)
print(f'Kolom dihapus: {drop_cols}')

# --- 6c. Filter kelembaban_tanah_1: nilai 0 tidak valid (sensor mati) ---
# Node 1 memiliki 26.6% nilai 0 → filter baris di mana SEMUA kelembaban tanah = 0
mask_semua_nol = (df['kelembaban_tanah_1'] == 0) & \
                 (df['kelembaban_tanah_2'] == 0) & \
                 (df['kelembaban_tanah_3'] == 0)
print(f'Baris semua sensor tanah = 0 (dihapus): {mask_semua_nol.sum()}')
df = df[~mask_semua_nol]

# --- 6d. Untuk kelembaban_tanah_1 = 0 saja → imputasi dengan interpolasi ---
df['kelembaban_tanah_1'] = df['kelembaban_tanah_1'].replace(0, np.nan)
before_interp = df['kelembaban_tanah_1'].isnull().sum()
df['kelembaban_tanah_1'] = df['kelembaban_tanah_1'].interpolate(method='time')
df['kelembaban_tanah_1'] = df['kelembaban_tanah_1'].fillna(method='bfill').fillna(method='ffill')
print(f'Nilai 0 pada kelembaban_tanah_1 diimputasi (interpolasi waktu): {before_interp}')

# --- 6e. Tekanan udara negatif → imputasi dengan median rolling ---
for col in ['tekanan_udara_1', 'tekanan_udara_2', 'tekanan_udara_3']:
    neg_mask = df[col] < 0
    if neg_mask.sum() > 0:
        df.loc[neg_mask, col] = np.nan
        df[col] = df[col].interpolate(method='linear').fillna(df[col].median())
        print(f'{col}: {neg_mask.sum()} nilai negatif → diimputasi')

# --- 6f. Curah hujan outlier → cap di 200mm (threshold realistis) ---
hujan_outlier = (df['curah_hujan'] > 200).sum()
df['curah_hujan'] = df['curah_hujan'].clip(upper=200)
print(f'curah_hujan di-cap 200mm: {hujan_outlier} nilai')

# --- 6g. Suhu dan kelembaban udara = 0 → imputasi ---
for col in ['suhu_udara_1', 'suhu_udara_2', 'suhu_udara_3',
            'kelembaban_udara_3']:
    zero_mask = df[col] == 0
    if zero_mask.sum() > 0:
        df.loc[zero_mask, col] = np.nan
        df[col] = df[col].interpolate(method='linear').fillna(df[col].median())
        print(f'{col}: {zero_mask.sum()} nilai nol → diimputasi')

print(f'\nBaris setelah cleaning: {len(df):,}')
print(f'Missing values tersisa : {df.isnull().sum().sum()}')
```

```python
# ============================================================
# CELL 7: Verifikasi Hasil Cleaning
# ============================================================
print('=== ✅ Statistik Data Setelah Cleaning ===')
print(df.describe().T[['mean', 'std', 'min', 'max']])
```

```python
# ============================================================
# CELL 8: Visualisasi Time Series Kelembaban Tanah
# ============================================================
fig, axes = plt.subplots(3, 1, figsize=(16, 10), sharex=True)
fig.suptitle('📈 Time Series Kelembaban Tanah per Node (Setelah Cleaning)', 
             fontsize=14, fontweight='bold')

colors = ['#1565C0', '#2E7D32', '#E65100']
for i, (col, color) in enumerate(zip(['kelembaban_tanah_1', 'kelembaban_tanah_2', 'kelembaban_tanah_3'], colors)):
    axes[i].plot(df['waktu_pembacaan'], df[col], color=color, linewidth=0.5, alpha=0.8)
    axes[i].set_ylabel('Kelembaban (%)', fontsize=10)
    axes[i].set_title(f'Node {i+1} – {col}', fontweight='bold')
    axes[i].set_ylim(0, 105)
    axes[i].axhline(y=60, color='red', linestyle='--', alpha=0.5, label='Threshold Kritis (60%)')
    axes[i].legend(fontsize=8)

axes[2].set_xlabel('Waktu Pembacaan', fontsize=10)
plt.tight_layout()
plt.savefig('dataset/plot_timeseries_cleaned.png', dpi=150, bbox_inches='tight')
plt.show()
print('✅ Plot disimpan ke dataset/plot_timeseries_cleaned.png')
```

```python
# ============================================================
# CELL 9: Simpan Dataset Bersih
# ============================================================
import os
os.makedirs('dataset', exist_ok=True)

df.to_csv('dataset/data_cleaned.csv', index=False)

print('=== 💾 Dataset Bersih Tersimpan ===')
print(f'File   : dataset/data_cleaned.csv')
print(f'Baris  : {len(df):,}')
print(f'Kolom  : {df.shape[1]}')
print(f'Kolom  : {list(df.columns)}')
print(f'Periode: {df["waktu_pembacaan"].min()} s.d. {df["waktu_pembacaan"].max()}')
print()
print('✅ Preprocessing selesai! Lanjut ke Notebook 02_preprocessing.ipynb')
```
